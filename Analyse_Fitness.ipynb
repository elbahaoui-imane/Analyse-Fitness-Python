```python
import pandas as pd

donnees = {
    'id_utilisateur':[1, 2, 3, 4, 5],
    'age':[18, 25, 30, 22, 35],
    'type_abonnement': ['Gratuit', 'Premium', 'Premium', 'Gratuit', 'Premium'],
    'prix_paye_euros': [0, 15, 15, 0, 15]
}

df = pd.DataFrame(donnees)

df
```

```python
chiffre_affaires = df['prix_paye_euros'].sum()
print("Chiffre d'affaires total :", chiffre_affaires, "Euros")

premium_only = df[df['type_abonnement'] == 'Premium']

age_moyen_premium = premium_only['age'].mean()
print("Age moyen des abonnes Premium :", age_moyen_premium, "ans")
```
